# Funktionen

* Ähnlich zu mathematischen Funktionen
$$y = f(x,t)$$

* Kleine Helfer\*innen und Expert\*innen für spezielle Aufgaben
  * Beispiel **Wurzelfunktion** $y = \sqrt{x}$ eines Taschenrechners zur Berechnung der Wurzel aus $x$

* Abstraktion und Wiederverwendtbarkeit
  * **Abstrahierung** der ggf. aufwendigen Rechenschritte z.B. zur Wurzelberechnung durch **sinnvolle Namensgebung**
  * **DRY**-Prinzip - **D**on't **R**epeat **Y**ourself

* Modularisierung
  * Aufbrechen komplexer Prozesse (mit sich wiederholenden Anweisungen) in kleinere Teilaufgaben
    * Im Gegensatz zu rein prozeduraler Programmierung
    * Gedankenexperiment: Sie kochen ein Hauptessen
      * [ ] Was können in diesem Gedankenbild alles Funktionen sein ?
  * Einfachere Wartung von Code

* **Black-Box**-Prinzip
  * Kapselung und Definition von geschützten Namensräumen
    * Eingaben und Aufgabe bekannt
      * $\sqrt{\cdot}$: Eingabezahl bekannt
    * Optional werden bestimmte Rückggaben erwartet
      * $\sqrt{\cdot}$: Erwarten Rückgabe **einer Zahl**
    * Keine Seiteneffekte: lokale Namensräume
      * Wiederverwendung von *Variablennamen* hat **keinen Einfluss** auf Programmstruktur
      * Kein Zugriff auf Variablen innerhalb einer Funktion
      * Definierte Schnittstellen für Datenflüssen in die Funktion und wieder heraus
        1. Übergabeparameter
        2. Rückgabeparameter
        3. globale Variablen

## Funktionsdefinition und Funktionsaufruf

* Eine Funktion muss definiert werden
  * Über- und Rückgabeparametern benennen
  
    ```Python
    def hallo_welt():
      print("Moin HAW!")
    ```
  * Funktionsbereich durch Einrückung definiert

* Der Funktionsname muss im aktuellen Namensraum - *Scope* verfügbar sein
  * ggf. muss die Funktion aus einem Paket importiert werden
  
    ```Python
    from math import cos
    ```

* Eine Funktion ruht bis zum Aufruf
  *
    ```Python
    def hallo_welt():
      print("Moin HAW!")
    
    hallo_welt()
    ```

### if

#### Aufbau

```
if <BEDINGUNG>:
  <AUSDRUCK>
  <AUSDRUCK>
  ...
```

* \<BEDINGUNG\>
  * Bool'scher- oder Python- Ausdruck mit Boole'scher Rückgabe
    * `True` oder `False`
* Der eingerückte Block nach `<BEDINGUNG>:` wird **NUR** ausgeführt, wenn `<BEDINGUNG>` wahr (`True`)

* \<AUSDRUCK\>
  * Eingerückte Zeile mit beliebigen, validen Python-Code


````python
bedingung = True
if bedingung:
  print("Bedingung ist Wahr!")
else:
  print("Falsch")
````

## pass, break, continue

* Schlüsselwörter für vordefinierte Kontrollflüsse


* `pass` - nützlich bei der aktiven Implementierungsarbeit
  * Beispiel Funktion definiert, aber ohne Inhalt

```Python
def macht_nichts():
  pass
```

* `break` und `continue` steuern den Kontrollfluss in Schleifen

### Aufgabe

* [ ] Erklären Sie die Wirkung von `continue` und `break` anhand folgenden Snippets

````python
from random import randint

geheime_zahl = randint(1,3)
bedingung = True

print("vor der Schleife")
while bedingung:
  print("Anfang")
  zahl = int(input("Geben Sie eine Ganzahl zwischen 1 und 3 ein"))
  
  if zahl < 1 or zahl > 3:
    print("Zahlen < 1 und > 3 sind nicht erlaubt.")
    continue
  
  if zahl == geheime_zahl:
    print("geheime Zahl gefunden!")
    break

print("Schleife beendet")
````
