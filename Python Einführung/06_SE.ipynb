````python
import matplotlib.pyplot as plt

def berechne_mittelwert(werte, name):
  laenge = len(werte)
  mittelwert = 0
  for index in range(laenge):
    mittelwert = mittelwert + werte[index]
  
  mittelwert = mittelwert/laenge

  
  plt.plot([i+1 for i in range(len(werte))], werte,'-x',label="x")
  plt.plot([1, len(werte)], [mittelwert,mittelwert],':o',label="<x>")
  plt.title("Werte mit Mittelwert")
  plt.xlabel("Index")
  plt.ylabel("Wert")
  plt.legend()
  
  plt.savefig(name);
  

berechne_mittelwert([0,0,0,1,1,1], "mittelwert.png")
````

### Zoom-In

````python
import numpy as np

def bearbeite_daten(daten):
  daten = daten - 5
  parameter_1 = 0.1232
  parameter_2 = 0.4
  Ny,Nx = daten.shape
  ergebnis = np.zeros((Ny,Nx))
  
  for i in range(Ny):
    for j in range(Nx):
      x = daten[i,j]
      if (parameter_1 < x and x < parameter_2):
        ergebnis[i,j] = x
      else:
        ergebnis[i,j] = 0
  return ergebnis

print(bearbeite_daten(np.random.rand(3,3)+5))
````

````python
import numpy as np

def ergebnis_ist_relevant(wert,p1,p2):
  return p1 < wert and wert < p2

def kernel(wert,p1,p2):
  if ergebnis_ist_relevant(wert,p1,p2):
    ergebnis = wert
  else:
    ergebnis = 0
  return ergebnis

def iteriere_ueber_daten(daten,p1,p2):
  Ny,Nx = daten.shape
  ergebnis = np.zeros((Ny,Nx))
  
  for i in range(Ny):
    for j in range(Nx):
      ergebnis[i,j] = kernel(daten[i,j], p1,p2)
  return ergebnis

def nivelliere_daten(daten):
  daten = daten - 5
  return daten

def filtere_daten(daten, p1=0.1232, p2=0.4):
  return iteriere_ueber_daten(daten,p1,p2)

def bearbeite_daten(daten):
  daten_niv = nivelliere_daten(daten)
  return filtere_daten(daten_niv)

print(bearbeite_daten(np.random.rand(3,3)+5))
````
