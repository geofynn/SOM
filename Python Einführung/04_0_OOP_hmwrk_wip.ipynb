````python
import matplotlib.pyplot as plt
%matplotlib widget
plt.ion()

class Graph(object):
  
  def __init__(self):
    self.objects =list()
  
  def plot(self, ax=None, fig=None):
    if ax is None:
      fig, ax = plt.subplots()
    for item in self.objects:
      item.plot(ax,fig)
    return fig, ax
  
  def clear(self,ax):
    for line in ax.lines:
      line.remove()

class Koordinate(object):
  meine_klassen_id = 1
  def __init__(self, x, y):
    self.instanz_id = Koordinate.meine_klassen_id
    Koordinate.meine_klassen_id +=1
    self.x = x
    self.y = y
  def verschiebe_position(self, dx=0, dy=0):
    self.x += dx
    self.y += dy
  def berechne_distanz(self,koordinate):
    dx2 = (self.x-koordinate.x)**2
    dy2 = (self.y-koordinate.y)**2
    return (dx2+dy2)**0.5

class GeometrieElement(object):
  
  def __init__(self,x,y,color='b',marker='x'):
    self.origin = Koordinate(x,y)
    self.path = dict()
    self.path['x'] = [x]
    self.path['y'] = [y]
    self.color = color
    self.marker = marker
  
  def verschiebe(self,dx=0,dy=0):
    self.origin.verschiebe_position(dx,dy)

  def plot(self,ax=None,fig=None):
    if ax is None:
      fig, ax = plt.subplots()
    ax.plot(self.path['x'],self.path['y'],'-',color=self.color,marker=self.marker)
    return fig, ax

class Rechteck(GeometrieElement):
  
  def __init__(self,x,y,a,b,color='b'):
    super().__init__(x,y,color=color,marker='')
    self.a = a
    self.b = b
    self.update_plot_path()
  
  def update_plot_path(self):
    x0 = self.origin.x
    y0 = self.origin.y
    self.path['x'] = [x0, x0+self.a, x0+self.a, x0, x0]
    self.path['y'] = [y0, y0, y0+self.b, y0+self.b, y0]
````
