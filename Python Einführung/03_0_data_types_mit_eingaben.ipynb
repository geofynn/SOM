## Formattierung von Strings

* Was sind *Strings* ? 
  * Vorerst nur Zeichenketten, die mit '"'-Zeichen umschlossen werden
  * Später mehr zu Strings ...

* Formattierte Ersetzungen in Strings durch **Platzhalter** u. `format()`-Methode

  *
    ```Python
    "Mein String zeigt am Ende eine Zahl:{}".format(3.14)
    ```
  * Analogon zu `fprintf`
### Beispiele

````python
s = "Mein String zeigt am Ende eine Zahl:{}".format(3.14)
print(s)
````

## Operationen auf sequenziellen Datentypen

* s,t: Instanzen eines seq. Datentyps (str, list, tuple ...)
* i,j,k,n: ganze Zahlen
* x: Referenz auf beliebige Instanz

|Nr |Ausdruck | Erklärung |
|---|---------|-----------|
| 1 |x in s   |Wahrheitsprüfung, ob x in s enthalten ist|
| 2 | s + t   | Verkettung von s mit t |
| 3 | s +=t   | Verkettung von s mit t, zugewiesen an s|
| 4 | n\*s oder s\*n | Neue Sequenz mit n verketteten Kopien von s|
| 5 | s\*= n  | wie 4, aber zugewiesen an s |
| 6 | s\[i\]  | i-te Element, **beginnend bei 0** |
| 7 | s\[i:j\]  | i-te bis j-te Element als Sequenz - **ausschließlich j** |
| 8 | s\[i:j:k\]  |  wie 7 aber in k-er Schritten als Sequenz |

Weitere Operationen

* `len(s)`
* `max(s)`
* `min(s)`
* `s.count(x)`
* `s.index(x,[i,j])`

## Listen

* Listen gruppieren (unterschiedliche) Datentypen in einem Objekt
  *
    ```Python
    meine_liste = [0,1,2,3,4]
    leere_liste = list()
    gemischte_liste = [0,"a",2]
    ```
  * Pro:
    * Sehr flexibel
  * Con:
    * Ressourcenaufwändig

* Zugriff auf Elemente der Liste mittels \[\]-Klammern
  *
      ```Python
      meine_liste = [0,1,2,3,4]
      print(meine_liste[1])
      print(meine_liste[4])
      ```
  * Indizierung beginnt bei **0**!
* Listen können verändert werden
  *
    ```Python
    meine_liste = [0,1,2,3,4]
    meine_liste[1] = 2 # Kein Fehler
    mein_str = "abc"
    mein_str[2] = "d" # Fehler!
    ```


### Beispiele

````python
meine_liste = [0,1,2,3,4]
leere_liste = list()
auch_leer = []
gemischte_liste = [0,"a",2]
print(meine_liste)
print(leere_liste)
print(type(auch_leer))
print(gemischte_liste)
````

  * `k in d`
  * `k not in d`
