## Initialisierung von Attributen

* Es ist möglich zu definieren *wie* eine Instanz erstellt wird
  * Mittels spezieller Methode `__init__`
  * `__` vorbehalten für spezielle Methoden (pseudo-privat)

````python
class Koordinate(object):
  def __init__(self, x,y):
    print("Der Konstruktor wird zuerst aufgerufen!")
    self.x = x
    self.y = y

# Objekt instansziieren
g = Koordinate(3,4)
h = Koordinate(-7,10)
print(g.x,g.y)
print(h.x,h.y)
````

## Methoden

* Analog zu `__init__(self,var1,...)`
* Analog Funktionen mit *zusätzlicher* Variablen `self` als 1. Übergabeparameter
  * Referenz/ Zugriffspunkt auf die *konkrete Instanz* der Klasse im Speicher (mit Zugriff auf *konkrete* Variablen)
  * Konvention `self` als ersten Parameternamen zu verwenden
* Zugriff über die *Instanz* + `.`-Operator  + *Methodenname*
  *
    ```Python
    g = Koordinate(3,4)
    dx, dy = (-3, 0)
    g.verschiebe_position(dx,dy)
    print(g.x,g.y)
    ```
* `self` wird beim Aufrufen der Methode nicht mitübergeben
  *
    ```Python
    g.verschiebe_position(dx,dy) # nur dx und dy übergeben
    ```

````python
class Koordinate(object):
  def __init__(self, x,y):
    print("Der Konstruktor wird zuerst aufgerufen!")
    self.x = x
    self.y = y
    self.verschiebe_position(10,10)
  
  def verschiebe_position(self, dx=0, dy=0):
    print("Dieser Text entsteht beim Aufrufen der Methode.")
    self.x += dx
    self.y += dy

# Objekt instansziieren
g = Koordinate(3,4)
m = Koordinate(5,6)
print(m.x,m.y)
# Zugriff auf Attribute
print(g.x,g.y)
# Zugriff auf Klassenmethoden
g.verschiebe_position(-3)
print(g.x,g.y)
g.x = 25
print(g.x,g.y)
````

## Erstellen und Verwenden von Klassen

* Man unterscheidet das **Erstellen einer Klasse** und die **Verwendung einer Instanz**
* **Erstellen** einer Klasse
  * Definieren des Klassennamens
  * Definieren von Klassenattributen und -methoden
  * z.B. Code für eine `list`-Klasse
* **Benutzen** einer Klasse
  * Erstellen und zuweisen von neuen **Instanzen** einer Klasse
  * Operationen mit den Attributen und Methoden der Instanz
  * z.B. `meine_liste = [1,2,3]` und `meine_liste.append(4)`

````python
meine_liste = [1,2,3]
meine_liste.append(4)
print(meine_liste)
````

* Bereits viele Standartobjekte von Python kennengelernt
  * Strings, Listen, Dictionaries, etc
* Klassen ermöglichen die Erstellung von **eigenen Datentypen**
  * Steigerung Komplexität

* **Klassendefinition**

````python
#class: Klassendefinition
#Koordinate: Name/Type
#(object): Elternklasse 'object'
class Koordinate(object):
  # Definition von Klassenattributen
  meine_klassen_id = 1
  # Definition von Klassenmethoden
  def __init__(self, x, y): # Der Konstruktor/ die Konstruktormethode
    # Definition und Verwendung von Instanz- und Klassenattributen  u.a.
    self.instanz_id = Koordinate.meine_klassen_id
    Koordinate.meine_klassen_id +=1
    self.x = x
    self.y = y
  def verschiebe_position(self, dx=0, dy=0): # Klassenmethode
    self.x += dx # Änderung Instanz-Variablen x um dx
    self.y += dy
````

* Schlüsselwort `class` wie `def`
  * Einrückung, um Codezuordnung zur Klasse zu definieren
* `Koordinate(object)` bedeutet, dass `Koordinate` ein Python-*Objekt* ist und entsprechend Attribute **erbt** (später mehr zu Vererbung)

* Spezielle Methode `__init__` - **Konstrutor**
  * Definiert (optional) die Realisierung eines Objektes/ Instanz
  * Initialisierung von Instanzattributen wie `instanz_id`
  * Parameter `self` zur Referenzierung der Instanz einer Klasse
    * Mit `self.` Zugriff auf Attribute und Methoden
  * `__` besteht aus 2 `_`

## Mehr zu Attributen und Methoden

* Daten und Funktionen, die zu der Klasse "gehören"
  * Definieren Objektverhalten

* Attribute
  * Andere Objekte, die Bausteine der Klasse sind
  * Beispiel `Koordinate` besteht aus zwei Zahlen

* Methoden
  * Funktionen, die nur im Zusammenhang mit der Klasse sinnvoll sind
  * Definieren Verhalten und Schnittstellen der Klasse
  * Realisieren zeitliche Entwicklung jeder Instanz
  * Beispiel wir können `Koordinate` *verschieben*, aber es macht keinen Sinn damit Listen zu verschieben
  

````python
class Koordinate(object):
  meine_klassen_id = 1
  def __init__(self, x, y):
    self.instanz_id = Koordinate.meine_klassen_id
    Koordinate.meine_klassen_id +=1
    self.x = x
    self.y = y
  def verschiebe_position(self, dx=0, dy=0):
    self.x += dx
    self.y += dy
  def berechne_distanz(self,koordinate):
    dx2 = (self.x-koordinate.x)**2
    dy2 = (self.y-koordinate.y)**2
    return (dx2+dy2)**0.5
  def __lt__(self,other):
    if isinstance(other,Koordinate):
      self.y = other.y 
      self.x = other.x
  def __str__(self):
   return f"x={self.x}" + f" y={self.y}"
````

````python
k1 = Koordinate(4,4)
k2 = Koordinate(-10,5)

print("Distanz zwischen beiden Koordinaten {:.2f}".format(k1.berechne_distanz(k2)))

k1 < k2
print(k1)
````

````python
class GeometrieElement(object):
  
  def __init__(self,x,y,color='b',marker='x'):
    print("Konstruktor GeometrieElement")
    self.origin = Koordinate(x,y)
    self.path = dict()
    self.path['x'] = [x]
    self.path['y'] = [y]
    self.color = color
    self.marker = marker
    print("Konstruktor GeometrieElement Ende")
  
  def verschiebe(self,dx=0,dy=0):
    self.origin.verschiebe_position(dx,dy)

  def plot(self,fig_no=1):
    fig = plt.figure(fig_no)
    plt.plot(self.path['x'],self.path['y'],'-',color=self.color,marker=self.marker)

class Rechteck(GeometrieElement):
  
  def __init__(self,x,y,a,b,color='b'):
    print("Konstruktor Rechteck")
    super().__init__(x,y,color=color,marker='')
    print("Konstruktor Rechteck Fortsetzung")
    self.a = a
    self.b = b
    self.update_plot_path()
    print("Konstruktor Rechteck Ende")
  
  def update_plot_path(self):
    x0 = self.origin.x
    y0 = self.origin.y
    self.path['x'] = [x0, x0+self.a, x0+self.a, x0, x0]
    self.path['y'] = [y0, y0, y0+self.b, y0+self.b, y0]

r1 = Rechteck(0,0,3,4)
r1.plot()
````
