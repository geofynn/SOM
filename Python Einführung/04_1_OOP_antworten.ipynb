# OOP Vorteile

*  **Datenbündelung in Paketen** 
  * Objekte, die gemeinsame Schnittstellen und Attributen teilen und damit eine wohldefinierte Interaktion erlauben
* **Devide-and-conquer** Entwicklung
  * Implementierung und Test des Verhaltens von jeder Klasse unabhängig
  * Erhöhung der Modularität verringert Komplexität
* Klassen vereinfachen **Wiederverwendung** von code
  * Viele Python Module definieren neue Klassen
  * Jede Klasse hat eine eigene Umgebung und Namensraum (keine Kollisionen mit Namen)
  * Vererbung erlaubt Kindklassen Methoden und Verhalten der Elternklassen zu spezialisieren und zu erweitern - *ohne die Elternklasse zu ändern*

* **Abstraktion**
  * Unterscheidung zwischen implementieren und verwenden von Objekten
  * Reduzierung auf das Wesentliche
* **Schichten** von Objektabstrahierung, die Verhalten von anderen Klassen erben
* **Eigene Klassen von Objekten** zu Pythons Standard-Klassen
