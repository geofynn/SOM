## Aufgabe OOP Deep-Dive: Setter, Getter ...

Getter sind dafür da um die Attribute in andere Programmteilen nutzen zu können.

Setter sind dafür da um die Attribute einer Klasse in anderen Programmteilen verändern zu können

Ein Unterstrich (_attribut): Dient als reiner Hinweis (Konvention). Es signalisiert: "Dieses Attribut ist intern, bitte greife nicht direkt von außen darauf zu." Technisch hält Python niemanden davon ab, es trotzdem zu tun (objekt._X funktioniert).

Zwei Unterstriche (__attribut): Löst das sogenannte Name Mangling (Namensverklausulierung) aus. Python ändert den internen Namen des Attributs automatisch in _Klassenname__attribut. Dies wird genutzt, um Namenskonflikte bei der Vererbung zu verhindern, damit Unterklassen nicht versehentlich Attribute der Oberklasse überschreiben.



````python
class A:
    def __init__(self):
        self._X = 100

    @property
    def X(self):
        return self._X
    
    @X.setter
    def X(self, new):
         if new < 0:
            print("Negativer Wert nicht erlaubt!")
            return
         else:
            self._X = new

Klasse = A()
print(Klasse.X)

Klasse.X = 10
print(Klasse.X)
````

hasattr(objekt, "attributname"): Prüft, ob ein Objekt ein bestimmtes Attribut oder eine Methode besitzt. Gibt True oder False zurück.

getattr(objekt, "attributname"): Liest den Wert eines Attributs dynamisch aus.

setattr(objekt, "attributname", wert): Setzt den Wert eines Attributs dynamisch (oder legt es neu an).


````python
obj = A()


if hasattr(obj, "_X"):
    aktueller_wert = getattr(obj, "_X")
    print(f"Gefundener Wert: {aktueller_wert}")

setattr(obj, "_X", 300)
print(obj.X)
````
