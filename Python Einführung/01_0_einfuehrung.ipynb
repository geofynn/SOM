## Legendre Polynome
Legendre Polynome : $P_n(x) = \frac {1}{2^{n}n!}\frac{d^n}{ dx^n}\left[(x^2 -1)^n\right]$

````python
import numpy as np # NumPy
import scipy as sp # SciPy
import matplotlib as mpl # Matplotlib (2D/3D)
import matplotlib.pyplot as plt # Matplotlib 's pyplot

x = np.linspace ( -1 ,1 ,100)
p1 = x
p2 = (3./2) *x**2 -1./2
p3 = (5./2) *x **3 -(3./2) *x
p4 = (35./8) *x**4 -(15./4)*x**2 + 3./8
plt.figure()
plt.plot(x,p1 ,'r:',x,p2 ,'g--',x,p3,x,p4,'m-',linewidth=2)
plt.xlabel( '$x$' , fontsize= 20)
plt.ylabel ( '$p_n(x)$' , fontsize =20)
plt.text( 0, 0.45 , 'Maximum ' )
plt.legend (('n=1', 'n=2', 'n=3', 'n=4'), loc='lower right')
plt.grid('on'), plt.box('on'), plt.xlim( (-1.1 , 1.1) )
````

## Mandelbrotmenge

Punkte $z\in\mathcal{C}$ für die die Folge

$z_0:= c$, $z_{n+1} = z^2_n + c$, $n\in\mathcal{N}$

endlich ist.

* Ist $|z_n|\geq 2$, dann divergiert die Folge (Stopp-Kriterium für die Berechnung)

````python
#import matplotlib.pyplot as plt
%matplotlib qt

def mandel():
    x1 = np.linspace (-2.1, 0.6, 3001)
    y1 = np.linspace (-1.1, 1.1, 2001)
    [X,Y] = np.meshgrid(x1 ,y1)
    it_max = 50
    Anz = np.zeros(X.shape) # shape means size
    C = (X + 1j*Y)          # C = X + Y i
    Z = np.copy(C)              # copy values
    for k in range(1,it_max):
     Z = Z**2+C
     Anz += np.isfinite(Z) # isfinite(A)=0 if A is not finite
    plt.figure()
    plt.imshow(Anz) # image(Anz)
    plt.colorbar()
mandel()
````
