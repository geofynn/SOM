```python
from math import pi
import matplotlib.pyplot as plt


class Circle(object):
    def __init__(self, x, y, r):
        self.x = x
        self.y = y
        self.r = r
    def area(self):
        area = pi*self.r**2
        return area
    def plot(self):
        fig, ax = plt.subplots()
        ax.set_xlim(-4, 4)
        ax.set_ylim(-4, 4)
        kreis = plt.Circle((self.x, self.y), radius=self.r, color='blue', fill=False, linewidth=2)
        ax.add_patch(kreis)
        plt.grid(True, linestyle='--')
        plt.show()
Kreis = Circle(2, 1, 2)



print(Kreis.area())
Kreis.plot()

import matplotlib.pyplot as plt
from matplotlib.patches import Wedge

fig, ax = plt.subplots()

# Wedge(center, r, theta1, theta2) - Winkel in Grad
semicircle = Wedge((0, 0), r=5, theta1=0, theta2=180, facecolor="lightblue", edgecolor="black")

ax.add_patch(semicircle)
ax.set_xlim(-6, 6)
ax.set_ylim(-1, 6)
ax.set_aspect("equal")
plt.show()
```
