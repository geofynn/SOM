$$\int^1_0 \exp(-x^2)dx$$

* [ ] Übersetzen Sie folgenden Matlab Code in Python Code:
    

* Importieren Sie das math-Modul für die Exponentialfunktion und verwenden Sie `for i in range(1,n):` für die Iteration von 1 bis n-1

````python
import math
from math import exp

n = 100
a = 0.0
b = 1.0
h = (b - a) / n
area = 0.5*h*( exp(-(a ** 2)) + exp(-(b ** 2)))
for i in range(1, n):
    x = a + i * h
    area = area + (h * exp(-(x ** 2)))

print('Der Wert der Approximation ist', area)
````
